```python
import random
import pandas as pd
import numpy as np
from pathlib import Path

# =========================
# CONFIGURATION
# =========================
TOTAL_SAMPLES = 24000
TRAIN_SIZE = 18000
TEST_SIZE = 3000
FINAL_TEST_SIZE = 3000

# Output folders 
TRAIN_DIR = Path("../../../data/training")
TEST_DIR = Path("../../../data/test")
FINAL_TEST_DIR = Path("../../../data/final test")

for d in (TRAIN_DIR, TEST_DIR, FINAL_TEST_DIR):
    d.mkdir(parents=True, exist_ok=True)

FEATURE_COLS = ["Temperature", "Pressure", "RPM", "Vibration"]

# =========================
# SAMPLE GENERATION (High Load)
# =========================
def gen_high_load_sample():
    temp = random.uniform(105, 115)        # Near upper safe limit
    rpm  = random.uniform(2500, 5500)      # Mid to high range
    pressure  = 0.9 + random.uniform(-0.07, 0.07)
    vibration = 0.10 + (rpm - 3500.0) / 20000.0 + random.uniform(-0.04, 0.05)
    return [temp, pressure, rpm, vibration], "High Load"

def build_dataset(n):
    X, y, rows = [], [], []
    seq_counter = 1
    for _ in range(n):
        features, label = gen_high_load_sample()
        X.append(features)
        y.append(label)
        rows.append({
            "Time": 1,
            "Sequence": seq_counter,
            "Temperature": features[0],
            "Pressure": features[1],
            "RPM": features[2],
            "Vibration": features[3],
            "State": label
        })
        seq_counter += 1
    # N -> (N, 1, 4)
    return np.array(X, dtype=np.float32).reshape(n, 1, 4), np.array(y), pd.DataFrame(rows)

# =========================
# GENERATE & SAVE SPLITS
# =========================
# Training
X_tr, y_tr, df_tr = build_dataset(TRAIN_SIZE)
df_tr.to_csv(TRAIN_DIR / "HighLoad_training.csv", index=False)
np.save(TRAIN_DIR / "HighLoad_training_X.npy", X_tr)
np.save(TRAIN_DIR / "HighLoad_training_y.npy", y_tr)

# Test
X_te, y_te, df_te = build_dataset(TEST_SIZE)
df_te.to_csv(TEST_DIR / "HighLoad_test.csv", index=False)
np.save(TEST_DIR / "HighLoad_test_X.npy", X_te)
np.save(TEST_DIR / "HighLoad_test_y.npy", y_te)

# Final test
X_ft, y_ft, df_ft = build_dataset(FINAL_TEST_SIZE)
df_ft.to_csv(FINAL_TEST_DIR / "HighLoad_final test.csv", index=False)
np.save(FINAL_TEST_DIR / "HighLoad_final test_X.npy", X_ft)
np.save(FINAL_TEST_DIR / "HighLoad_final test_y.npy", y_ft)

print("✅ High Load datasets created.")
print("Training:", X_tr.shape, y_tr.shape)
print("Test:    ", X_te.shape, y_te.shape)
print("Final:   ", X_ft.shape, y_ft.shape)
```
