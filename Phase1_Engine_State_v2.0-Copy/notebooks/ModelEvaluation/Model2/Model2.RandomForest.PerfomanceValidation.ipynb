```python
import numpy as np
import joblib
from sklearn.metrics import accuracy_score, classification_report

# -----------------------------
# Config
# -----------------------------
MODEL_PATH  = "../../../models/Model2/model2_randomforest.pkl"
X_TEST_PATH = "../../../data/final test/EngineOff_final test_X.npy"
Y_TEST_PATH = "../../../data/final test/EngineOff_final test_y.npy"

# -----------------------------
# Load data
# -----------------------------
X_test = np.load(X_TEST_PATH)
y_raw  = np.load(Y_TEST_PATH, allow_pickle=True)  # may be strings

# Flatten to 2D (handles shapes like (N,1,1))
X_test = X_test.reshape(X_test.shape[0], -1)

# Map labels if they are strings
label_map = {"Engine Off (cold)": 0, "Engine Off (warm)": 1}
if y_raw.dtype.kind in {"U", "S", "O"}:  # strings/objects
    y_test = np.array([label_map[str(v)] for v in y_raw], dtype=int)
    target_names = list(label_map.keys())
else:
    y_test = y_raw.astype(int)
    target_names = ["Engine Off (cold)", "Engine Off (warm)"]

# -----------------------------
# Load & evaluate
# -----------------------------
rf = joblib.load(MODEL_PATH)
y_pred = rf.predict(X_test)

acc = accuracy_score(y_test, y_pred)
print(f"✅ Final Test Accuracy: {acc:.4f}")
print("\nClassification Report:\n",
      classification_report(y_test, y_pred, target_names=target_names))
```
