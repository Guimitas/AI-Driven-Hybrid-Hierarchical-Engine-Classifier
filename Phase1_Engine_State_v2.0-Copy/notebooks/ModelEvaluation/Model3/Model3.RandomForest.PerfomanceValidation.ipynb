```python
import numpy as np
import joblib
from sklearn.metrics import accuracy_score, classification_report

# -----------------------------
# Config
# -----------------------------
MODEL_PATH = "../../../models/Model3/model3_randomforest.pkl"

base = "../../../data/final test/"
X_paths = [
    base + "NormalLoad_final test_X.npy",
    base + "HighLoad_final test_X.npy",
    base + "CriticalLoad_final test_X.npy",
]
y_paths = [
    base + "NormalLoad_final test_y.npy",
    base + "HighLoad_final test_y.npy",
    base + "CriticalLoad_final test_y.npy",
]

# -----------------------------
# Load & concat
# -----------------------------
X_list = [np.load(p, allow_pickle=True) for p in X_paths]
y_list = [np.load(p, allow_pickle=True) for p in y_paths]

X_test = np.concatenate(X_list, axis=0)
y_raw  = np.concatenate(y_list, axis=0)

# Flatten
X_test = X_test.reshape(X_test.shape[0], -1)

# -----------------------------
# Map labels
# -----------------------------
def to_int_label(lbl):
    if isinstance(lbl, (int, np.integer)):
        return int(lbl)
    s = str(lbl).lower().replace(" ", "").replace("_", "")
    if "normal" in s:
        return 0
    if "high" in s:
        return 1
    if "critical" in s:
        return 2
    raise ValueError(f"Unrecognized label: {lbl}")

y_test = np.array([to_int_label(v) for v in y_raw], dtype=int)
target_names = ["NormalLoad (0)", "HighLoad (1)", "CriticalLoad (2)"]

# -----------------------------
# Load & evaluate
# -----------------------------
payload = joblib.load(MODEL_PATH)
rf = payload["model"]   # model stored
y_pred = rf.predict(X_test)

acc = accuracy_score(y_test, y_pred)
print(f"✅ Final Test Accuracy: {acc:.4f}")
print("\nClassification Report:\n",
      classification_report(y_test, y_pred, target_names=target_names, digits=4))
```
