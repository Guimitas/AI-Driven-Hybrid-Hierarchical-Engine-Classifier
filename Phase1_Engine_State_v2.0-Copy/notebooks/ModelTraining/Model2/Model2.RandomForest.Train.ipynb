```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
import joblib

# ==== 1. Load the data ====
X_path = "../../../data/training/EngineOff_training_X.npy"
y_path = "../../../data/training/EngineOff_training_y.npy"

X = np.load(X_path)
y = np.load(y_path)

# ==== 2. Flatten the input from (N, 1, 1) → (N, 1) ====
X = X.reshape(X.shape[0], -1)

# ==== 3. Map labels to integers ====
label_map = {
    "Engine Off (cold)": 0,
    "Engine Off (warm)": 1
}
y_mapped = np.array([label_map[label] for label in y])

# ==== 4. Train RandomForest ====
rf_model = RandomForestClassifier(
    n_estimators=100,     # number of trees
    random_state=42,      # reproducibility
    n_jobs=-1             # use all CPU cores
)
rf_model.fit(X, y_mapped)

# ==== 5. Evaluate ====
y_pred = rf_model.predict(X)
print("Training Accuracy:", accuracy_score(y_mapped, y_pred))
print("\nClassification Report:\n", classification_report(y_mapped, y_pred, target_names=label_map.keys()))

# ==== 6. Save the model ====
save_path = "../../../models/Model2/model2_randomforest.pkl"
joblib.dump(rf_model, save_path)
print(f"Model saved to: {save_path}")
```
